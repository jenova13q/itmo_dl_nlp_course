```python
from corus import load_lenta
import pandas as pd
import numpy as np
import re
import random

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

path = 'lenta-ru-news.csv.gz'  # путь к файлу, например, в локальном колабе
records = load_lenta(path)

# Чтобы не работать со всем датасетом (он достаточно большой ~1 млн новостей),
# возьмём репрезентативную часть, например, 100_000 записей.
# Заодно сразу будем собирать нужные поля (title, text, topic).
data = []
MAX_SIZE = 100000
for i, record in enumerate(records):
    # Иногда бывает None в text, topic: надо будет отбросить
    if record.text is not None and record.topic is not None:
        data.append({
            'title': record.title,
            'text': record.text,
            'topic': record.topic
        })
    if len(data) >= MAX_SIZE:
        break

df = pd.DataFrame(data)
print(df.shape)
df.head()
```

```python
import pymorphy3
from tqdm import tqdm
tqdm.pandas()

morph = pymorphy3.MorphAnalyzer()

def clean_text(text):
    # 1. Приводим к нижнему регистру
    text = text.lower()
    # 2. Удаляем всё, кроме букв и пробелов
    text = re.sub(r'[^а-яёa-z\s]', ' ', text)
    # 3. Разбиваем на токены (примитивно по пробелам)
    tokens = text.split()
    # 4. Лемматизируем
    lemmas = []
    for token in tokens:
        p = morph.parse(token)[0]
        lemmas.append(p.normal_form)
    # 5. Склеиваем обратно
    return " ".join(lemmas)

# Для ускорения возьмём, скажем, только текст.
df["clean_text"] = df["text"].progress_apply(clean_text)
```

```python
counts = df["topic"].value_counts()
rare_topics = counts[counts < 2].index  # классы, в которых <2 примеров
df = df[~df["topic"].isin(rare_topics)]
```

```python
from sklearn.model_selection import train_test_split

# Делим 80/20
train_val_df, test_df = train_test_split(df,
                                         test_size=0.2,
                                         shuffle=True,
                                         random_state=RANDOM_SEED,
                                         stratify=df["topic"])

# Теперь делим train_val ещё на 75/25 => 0.75 * 0.8 = 0.6 от всего
train_df, val_df = train_test_split(train_val_df,
                                    test_size=0.25,  # (это 25% от 80%)
                                    shuffle=True,
                                    random_state=RANDOM_SEED,
                                    stratify=train_val_df["topic"])

print("Train:", train_df.shape)
print("Val:  ", val_df.shape)
print("Test: ", test_df.shape)

X_train, y_train = train_df["clean_text"], train_df["topic"]
X_val, y_val     = val_df["clean_text"],   val_df["topic"]
X_test, y_test   = test_df["clean_text"],  test_df["topic"]
```

# Замерьте базовое качество с любым dummy-бейзлайном - 0.5 балла

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

dummy = DummyClassifier(strategy='most_frequent', random_state=RANDOM_SEED)
dummy.fit(X_train, y_train)  # fit "как бы" учится, но реально выдаёт константу
y_pred_val = dummy.predict(X_val)
print("Dummy accuracy:", accuracy_score(y_val, y_pred_val))
```

#Обучите модель sklearn.linear_model.LogisticRegression с двумя вариантами векторизации: 2 балла
*   sklearn.feature_extraction.text.CountVectorizer
*   sklearn.feature_extraction.text.TfidfVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
pipe_count = Pipeline([
    ("vect", CountVectorizer()),  # векторизация
    ("clf",  LogisticRegression(random_state=RANDOM_SEED, max_iter=1000))
])

pipe_count.fit(X_train, y_train)
y_pred_val_count = pipe_count.predict(X_val)

acc_count = accuracy_score(y_val, y_pred_val_count)
print("LogReg + CountVectorizer. Accuracy на валидации:", acc_count)
```

```python
y_pred_test_count = pipe_count.predict(X_test)
acc_test_count = accuracy_score(y_test, y_pred_test_count)
print("Accuracy на тесте:", acc_test_count)
```

```python
pipe_tfidf = Pipeline([
    ("vect", TfidfVectorizer()),
    ("clf",  LogisticRegression(random_state=RANDOM_SEED, max_iter=1000))
])

pipe_tfidf.fit(X_train, y_train)
y_pred_val_tfidf = pipe_tfidf.predict(X_val)

acc_tfidf = accuracy_score(y_val, y_pred_val_tfidf)
print("LogReg + TfidfVectorizer. Accuracy на валидации:", acc_tfidf)
```

```python
y_pred_test_tfidf = pipe_tfidf.predict(X_test)
acc_test_tfidf = accuracy_score(y_test, y_pred_test_tfidf)
print("Accuracy на тесте:", acc_test_tfidf)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "vect__ngram_range": [(1, 1)],
    "vect__max_df": [0.9, 0.95],
    "vect__min_df": [1, 2],
    "clf__C": [1.0, 10.0]
}

pipe_for_search = Pipeline([
    ("vect", TfidfVectorizer()),
    ("clf",  LogisticRegression(random_state=42, max_iter=1000))
])

gs = GridSearchCV(
    estimator=pipe_for_search,
    param_grid=param_grid,
    scoring="accuracy",
    cv=3,
    verbose=1,
    n_jobs=1
)

gs.fit(X_train, y_train)

print("Лучшие параметры:", gs.best_params_)
print("Лучшая точность на кросс-валидации:", gs.best_score_)
```

# Обоснование принятых решений (1 балл)
## Выбор лемматизации и удаления «шума»:

Лемматизация русского языка (через pymorphy3) позволяет свести склонения и спряжения к нормальной форме, уменьшая размер словаря и потенциальный «шум».
Удаление ссылок, цифр и т.п. — чтобы не засорять модель бесполезными признаками.

## Ограничение набора до 100k:

Для снижения времени обучения и предобработки. Этой выборки достаточно репрезентативно.

## Сплит 60/20/20:

Стандартный приём, когда хочется иметь и валидацию, и тест, причём в равных пропорциях для всех классов (стратификация).
## Бейзлайн:

DummyClassifier (most_frequent) показывает, какова точность, если модель игнорирует входные данные и выдаёт самый популярный класс. Нужно, чтобы оценить пользу «умной» модели.
## Почему LogisticRegression:

Классическая «линейная» модель для текстов, хорошо работает, даёт интерпретацию по весам признаков, быстро обучается, особенно на разреженных данных.
## Почему CountVectorizer + TfidfVectorizer:

Два простых и распространённых подхода «мешка слов»: Bag-of-Words vs TF-IDF. TF-IDF учитывает важность слова в документе с учётом частоты встречаемости в корпусе.
## Подбор гиперпараметров:

Для векторизации (размер n-грамм, фильтрация частых/редких слов) и для самой логистической регрессии (C) — стандартный шаг для улучшения качества.
